## Проект "Деревья решений и Метод опорных векторов"

### Раздел 1. Описание набора данных

Для проекта используется набор данных о бронированиях двух типов отелей: курортного (Resort Hotel) и городского (City Hotel). Каждая строка соответствует отдельному бронированию и содержит сведения о сроках проживания, составе гостей, условиях бронирования и его итоговом статусе.

Набор данных включает 119 390 наблюдений и 32 столбца. В нём представлены количественные и качественные признаки. К количественным относятся срок между бронированием и заездом (lead_time), количество ночей в выходные и будние дни (stays_in_weekend_nights, stays_in_week_nights), число взрослых, детей и младенцев (adults, children, babies), а также количество специальных запросов (total_of_special_requests). К качественным признакам относятся тип отеля (hotel), тип питания (meal), страна гостя (country), сегмент рынка (market_segment), канал бронирования (distribution_channel) и тип забронированного номера (reserved_room_type).

В рамках проекта рассматриваются две задачи машинного обучения. Для классификации целевой переменной является is_canceled: значение 1 обозначает отменённое бронирование, а 0 — бронирование, которое не было отменено. Задача состоит в прогнозировании вероятного факта отмены по характеристикам бронирования. Для регрессии целевой переменной выступает adr — средняя суточная стоимость проживания. В этом случае необходимо предсказать её числовое значение.

```python
monthly_adr = (
    monthly_summary['adr_mean']
    .unstack('hotel')
    .reindex(month_order)
)

ax = monthly_adr.plot(
    marker='o',
    figsize=(12, 5)
)

ax.set_xticks(range(12))
ax.set_xticklabels(month_labels)
ax.set(
    title='Средний ADR по месяцу заезда и типу отеля',
    xlabel='Месяц заезда',
    ylabel='Средний ADR'
)
ax.legend(title='Тип отеля')

plt.tight_layout()
plt.show()

display(monthly_adr.round(2))
```

### Раздел 4.	Подготовка данных к моделированию

```python
# Сохраняем исходный df для воспроизводимости анализа
data = df.copy()

print('Исходный размер:', data.shape)
print('Полных дубликатов:', data.duplicated().sum())
```

```python
# Общее количество ночей
total_nights = (
    data['stays_in_weekend_nights']
    + data['stays_in_week_nights']
)

# min_count=3: сумма считается только при наличии всех трёх значений
total_guests = data[['adults', 'children', 'babies']].sum(
    axis=1, min_count=3
)

checks = pd.Series({
    'Отрицательный ADR': data['adr'].lt(0).sum(),
    'Нулевой ADR': data['adr'].eq(0).sum(),
    'Нет гостей': total_guests.eq(0).sum(),
    'Нет ночей проживания': total_nights.eq(0).sum()
})

display(checks.to_frame('Количество строк'))

print('\nБронирования с отрицательным ADR:')
display(
    data.loc[
        data['adr'].lt(0),
        [
            'hotel', 'adults', 'children', 'babies',
            'stays_in_weekend_nights', 'stays_in_week_nights',
            'adr', 'is_canceled'
        ]
    ]
)
```

Проверка выявила одну строку с отрицательным ADR, 1 959 строк с нулевой стоимостью, 180 бронирований без гостей и 715 — без ночей проживания. Эти группы могут пересекаться, поэтому складывать их количества нельзя.

Для проекта ограничим выборку бронированиями хотя бы с одним гостем и одной ночью проживания, а отрицательный ADR исключим как некорректную стоимость. Нулевой ADR пока сохраним: без дополнительной информации его нельзя считать ошибкой.

```python
# Удаляем полные дубликаты
rows_before = len(data)
data = data.drop_duplicates().copy()

print('Удалено полных дубликатов:', rows_before - len(data))

# Повторно рассчитываем показатели после удаления дубликатов
total_guests = data[['adults', 'children', 'babies']].sum(
    axis=1, min_count=3
)

total_nights = (
    data['stays_in_weekend_nights']
    + data['stays_in_week_nights']
)

# Исключаем строки, не соответствующие выбранной области исследования
exclude = (
    data['adr'].lt(0)
    | total_guests.eq(0)
    | total_nights.eq(0)
)

print('Дополнительно исключено строк:', exclude.sum())

data = data.loc[~exclude].copy()

print('Размер подготовленной таблицы:', data.shape)
print('Осталось нулевых значений ADR:', data['adr'].eq(0).sum())
```

После удаления 31 994 полных дубликатов и дополнительного исключения 758 наблюдений в выборке осталось 86 638 строк и 32 столбца. Сохранены 1 052 бронирования с нулевым ADR, поскольку нулевая стоимость сама по себе не подтверждает ошибку.

Выбор признаков и целевых переменных

Для обеих задач используем общий набор входных признаков. Исключим:
- reservation_status и reservation_status_date — сведения об итоговом состоянии бронирования;
- assigned_room_type и booking_changes — сведения, которые могут появляться или изменяться после оформления бронирования;

```python
drop_columns = [
    'is_canceled',
    'adr',
    'reservation_status',
    'reservation_status_date',
    'assigned_room_type',
    'booking_changes',
    'agent',
    'company'
]

X = data.drop(columns=drop_columns)

y_class = data['is_canceled']
y_reg = data['adr']

print('Количество входных признаков:', X.shape[1])
print('Признаки:', X.columns.tolist())
```

```python
from sklearn.model_selection import train_test_split

train_idx, test_idx = train_test_split(
    X.index,
    test_size=0.2,
    random_state=42,
    stratify=y_class
)

X_train = X.loc[train_idx].copy()
X_test = X.loc[test_idx].copy()

y_train_class = y_class.loc[train_idx].copy()
y_test_class = y_class.loc[test_idx].copy()

y_train_reg = y_reg.loc[train_idx].copy()
y_test_reg = y_reg.loc[test_idx].copy()

print('Обучающая выборка:', X_train.shape)
print('Тестовая выборка:', X_test.shape)

class_proportions = pd.DataFrame({
    'Вся очищенная выборка, %': y_class.value_counts(normalize=True) * 100,
    'Train, %': y_train_class.value_counts(normalize=True) * 100,
    'Test, %': y_test_class.value_counts(normalize=True) * 100
}).sort_index()

display(class_proportions.round(2))
```

В обучающую выборку вошло 69 310 наблюдений, в тестовую — 17 328, при 24 входных признаках. Стратификация сохранила пропорции классов: около 72,31% неотменённых и 27,69% отменённых бронирований.

```python
#Разделение признаков по типам
numeric_features = X_train.select_dtypes(
    include='number'
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=['object', 'category']
).columns.tolist()

print('Числовых признаков:', len(numeric_features))
print(numeric_features)

print('\nКатегориальных признаков:', len(categorical_features))
print(categorical_features)
```

```python
#Предобработка для деревьев и случайного леса
# Числовые пропуски заполним медианой, категориальные — отдельным значением Unknown. Категории закодируем методом One-Hot Encoding.
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor_tree = ColumnTransformer(
    transformers=[
        (
            'num',
            SimpleImputer(strategy='median'),
            numeric_features
        ),
        (
            'cat',
            Pipeline([
                ('imputer', SimpleImputer(
                    strategy='constant',
                    fill_value='Unknown'
                )),
                ('encoder', OneHotEncoder(handle_unknown='ignore'))
            ]),
            categorical_features
        )
    ]
)

print('Предобработка для деревьев подготовлена.')
```

```python
# Предобработка для SVM и SVR
# Для этих моделей дополнительно стандартизируем числовые признаки. Бинарные столбцы, полученные после One-Hot Encoding, оставим без стандартизации.
preprocessor_svm = ColumnTransformer(
    transformers=[
        (
            'num',
            Pipeline([
                ('imputer', SimpleImputer(strategy='median')),
                ('scaler', StandardScaler())
            ]),
            numeric_features
        ),
        (
            'cat',
            Pipeline([
                ('imputer', SimpleImputer(
                    strategy='constant',
                    fill_value='Unknown'
                )),
                ('encoder', OneHotEncoder(handle_unknown='ignore'))
            ]),
            categorical_features
        )
    ]
)

print('Предобработка для SVM и SVR подготовлена.')
```

Теперь настроим балансировку классов. Используем RandomOverSampler: при обучении он будет дублировать наблюдения класса меньшинства до равного количества классов.

```python
from imblearn.over_sampling import RandomOverSampler
from imblearn.pipeline import Pipeline as ImbPipeline

sampler = RandomOverSampler(random_state=42)

print('Распределение классов в обучающей выборке:')
display(
    y_train_class.value_counts()
    .sort_index()
    .rename_axis('is_canceled')
    .to_frame('Количество')
)

print('Балансировщик подготовлен.')
```

Подготовлены отдельные схемы предобработки для деревьев и методов опорных векторов. Для SVM и SVR предусмотрена стандартизация числовых признаков. Для классификации настроена балансировка методом случайного увеличения класса меньшинства, которая будет выполняться внутри обучающего конвейера. Для регрессии балансировка классов не применяется.

### Раздел 5. Построение DecisionTreeClassifier

```python
# Конвейер и кросс-валидация
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV, StratifiedKFold

cv_class = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)

tree_pipeline = ImbPipeline([
    ('preprocessor', preprocessor_tree),
    ('sampler', RandomOverSampler(random_state=42)),
    ('model', DecisionTreeClassifier(random_state=42))
])
```

Здесь мы подготовили схему обучения дерева решений, но ещё не обучили модель.

cv_class задаёт кросс-валидацию: обучающая выборка делится на 3 части с сохранением пропорций классов. Модель будет обучаться на двух частях и проверяться на третьей, поочерёдно меняя их роли. shuffle=True перемешивает строки, а random_state=42 делает разбиение воспроизводимым.

```python
# Подбор гиперпараметров
tree_param_grid = {
    'model__max_depth': [5, 10, 15, 20],
    'model__min_samples_leaf': [5, 20, 50],
    'model__criterion': ['gini', 'entropy']
}

tree_search = GridSearchCV(
    estimator=tree_pipeline,
    param_grid=tree_param_grid,
    scoring={
        'f1': 'f1',
        'accuracy': 'accuracy',
        'balanced_accuracy': 'balanced_accuracy'
    },
    refit='f1',
    cv=cv_class,
    n_jobs=-1,
    return_train_score=True,
    verbose=1
)

tree_search.fit(X_train, y_train_class)

print('Лучшие параметры:')
print(tree_search.best_params_)

print('\nСредний F1 на кросс-валидации:',
      round(tree_search.best_score_, 4))
```

```python
# Оценка модели
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

best_tree = tree_search.best_estimator_

tree_train_pred = best_tree.predict(X_train)
tree_test_pred = best_tree.predict(X_test)

def classification_metrics(y_true, y_pred):
    return {
        'Accuracy': accuracy_score(y_true, y_pred),
        'Balanced accuracy': balanced_accuracy_score(y_true, y_pred),
        'Precision': precision_score(y_true, y_pred, zero_division=0),
        'Recall': recall_score(y_true, y_pred, zero_division=0),
        'F1': f1_score(y_true, y_pred, zero_division=0)
    }

tree_metrics = pd.DataFrame({
    'Train': classification_metrics(y_train_class, tree_train_pred),
    'Test': classification_metrics(y_test_class, tree_test_pred)
})

display(tree_metrics.round(4))

# Ориентир: всегда предсказывать класс 0
print(
    'Тестовая accuracy при постоянном прогнозе «не отменено»:',
    round(y_test_class.eq(0).mean(), 4)
)

# Разница между обучением и валидацией для выбранных параметров
best_index = tree_search.best_index_

print(
    '\nСредний F1 на обучающих фолдах:',
    round(tree_search.cv_results_['mean_train_f1'][best_index], 4)
)
print(
    'Средний F1 на валидационных фолдах:',
    round(tree_search.best_score_, 4)
)
```

По результатам GridSearchCV выбрано дерево с критерием gini, максимальной глубиной 15 и минимальным количеством наблюдений в листе 50. На тестовой выборке accuracy составила 76,37%, что превышает требуемые 70% и результат постоянного прогноза «не отменено» — 72,32%. Модель выявляет 80,88% отменённых бронирований, однако precision равна 54,97%, то есть среди прогнозируемых отмен остаётся значительная доля ложных срабатываний. F1 составляет 0,6546.

```python
from sklearn.metrics import ConfusionMatrixDisplay, confusion_matrix

tree_cm = confusion_matrix(
    y_test_class,
    tree_test_pred,
    labels=[0, 1]
)

ConfusionMatrixDisplay(
    confusion_matrix=tree_cm,
    display_labels=['Не отменено', 'Отменено']
).plot(cmap='Blues', values_format='d')

plt.title('Матрица ошибок DecisionTreeClassifier')
plt.xlabel('Предсказанный класс')
plt.ylabel('Истинный класс')
plt.tight_layout()
plt.show()
```

Модель правильно определила 9 353 неотменённых и 3 880 отменённых бронирований. При этом она пропустила 917 отмен и ошибочно предсказала отмену для 3 178 неотменённых бронирований. Таким образом, дерево выявляет большинство отмен

### Раздел 6. «Построение RandomForestClassifier»

Предобработка и балансировка выполняются внутри конвейера. Параметр регуляризации C подберём с помощью GridSearchCV по метрике F1, используя прежние обучающую и тестовую выборки и схему кросс-валидации.

### Раздел 8. Построение DecisionTreeRegressor

Для прогнозирования ADR построим дерево решений для регрессии. Используем ранее зафиксированные обучающую и тестовую выборки. Подбор гиперпараметров выполним через GridSearchCV по MAE — средней абсолютной ошибке. Дополнительно оценим RMSE и R². Стандартизация для дерева не требуется, балансировка классов в задаче регрессии не применяется.

```python
# Конвейер и кросс-валидация
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import KFold

cv_reg = KFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)

reg_tree_pipeline = Pipeline([
    ('preprocessor', preprocessor_tree),
    ('model', DecisionTreeRegressor(random_state=42))
])
```

```python
reg_tree_param_grid = {
    'model__max_depth': [5, 10, 15, 20],
    'model__min_samples_leaf': [5, 20, 50]
}

reg_tree_search = GridSearchCV(
    estimator=reg_tree_pipeline,
    param_grid=reg_tree_param_grid,
    scoring='neg_mean_absolute_error',
    cv=cv_reg,
    n_jobs=-1,
    pre_dispatch='n_jobs',
    return_train_score=True,
    verbose=1
)

reg_tree_search.fit(X_train, y_train_reg)

print('Лучшие параметры:')
print(reg_tree_search.best_params_)

print('\nСредняя MAE на кросс-валидации:',
      round(-reg_tree_search.best_score_, 4))
```

Лучшей из проверенных комбинаций оказалось дерево с максимальной глубиной 20 и минимумом 5 наблюдений в листе. Средняя MAE на кросс-валидации составила 14,4047: прогноз ADR отличается от фактического значения в среднем примерно на 14,40 единицы.

```python
# Оценка модели
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

def regression_metrics(y_true, y_pred):
    return {
        'MAE': mean_absolute_error(y_true, y_pred),
        'RMSE': mean_squared_error(y_true, y_pred) ** 0.5,
        'R²': r2_score(y_true, y_pred)
    }

best_reg_tree = reg_tree_search.best_estimator_

reg_tree_train_pred = best_reg_tree.predict(X_train)
reg_tree_test_pred = best_reg_tree.predict(X_test)

reg_tree_metrics = pd.DataFrame({
    'Train': regression_metrics(y_train_reg, reg_tree_train_pred),
    'Test': regression_metrics(y_test_reg, reg_tree_test_pred)
})

display(reg_tree_metrics.round(4))

best_index = reg_tree_search.best_index_

print(
    'Средняя MAE на обучающих фолдах:',
    round(
        -reg_tree_search.cv_results_['mean_train_score'][best_index],
        4
    )
)
print(
    'Средняя MAE на валидационных фолдах:',
    round(-reg_tree_search.best_score_, 4)
)
```

На тестовой выборке дерево регрессии достигло MAE 13,32, RMSE 21,35 и R² 0,8274. Это означает, что средняя абсолютная ошибка прогноза ADR составляет около 13,32 единицы, а модель объясняет примерно 82,74% вариации стоимости на тесте. MAE на обучении ниже — 9,32, что указывает на признаки переобучения. При этом тестовая ошибка близка к результату кросс-валидации — 14,40, что подтверждает способность модели прогнозировать новые наблюдения.

```python
# Сравнение с простым прогнозом
from sklearn.dummy import DummyRegressor

dummy_reg = DummyRegressor(strategy='median')

dummy_reg.fit(X_train, y_train_reg)
dummy_reg_pred = dummy_reg.predict(X_test)

reg_comparison = pd.DataFrame({
    'Прогноз медианой': regression_metrics(y_test_reg, dummy_reg_pred),
    'DecisionTreeRegressor': reg_tree_metrics['Test']
})

display(reg_comparison.round(4))
```

Дерево регрессии значительно превосходит постоянный прогноз медианой: MAE снизилась с 38,86 до 13,32, а RMSE — с 52,00 до 21,35. Значение R² выросло с −0,0242 до 0,8274. Это подтверждает, что модель использует признаки для прогнозирования ADR существенно эффективнее простого базового подхода.

### Раздел 9. Построение SVR

Для прогнозирования ADR применим линейную регрессию на основе метода опорных векторов — LinearSVR. Числовые признаки стандартизируем, категориальные закодируем методом One-Hot Encoding. Параметры C и epsilon подберём через GridSearchCV по средней абсолютной ошибке. Для сравнения с деревом регрессии используем прежние выборки и фолды кросс-валидации. Балансировка в задаче регрессии не применяется.

```python
#Создание конвейера
from sklearn.svm import LinearSVR

svr_pipeline = Pipeline([
    ('preprocessor', preprocessor_svm),
    ('model', LinearSVR(
        loss='squared_epsilon_insensitive',
        dual=False,
        max_iter=20000,
        random_state=42
    ))
])
```

```python
# Подбор гиперпараметров
svr_param_grid = {
    'model__C': [0.1, 1.0, 10.0],
    'model__epsilon': [0.0, 1.0]
}

svr_search = GridSearchCV(
    estimator=svr_pipeline,
    param_grid=svr_param_grid,
    scoring='neg_mean_absolute_error',
    cv=cv_reg,
    n_jobs=-1,
    pre_dispatch='n_jobs',
    return_train_score=True,
    verbose=2
)

svr_search.fit(X_train, y_train_reg)

print('Лучшие параметры:')
print(svr_search.best_params_)

print('\nСредняя MAE на кросс-валидации:',
      round(-svr_search.best_score_, 4))
```

Лучшими из проверенных параметров LinearSVR стали C = 1,0 и epsilon = 0,0. Средняя MAE на кросс-валидации составила 22,44, что выше ошибки дерева регрессии — 14,40. Следовательно, на кросс-валидации дерево точнее прогнозирует ADR.

```python
# Оценка качества
best_svr = svr_search.best_estimator_

svr_train_pred = best_svr.predict(X_train)
svr_test_pred = best_svr.predict(X_test)

svr_metrics = pd.DataFrame({
    'Train': regression_metrics(y_train_reg, svr_train_pred),
    'Test': regression_metrics(y_test_reg, svr_test_pred)
})

display(svr_metrics.round(4))

best_index = svr_search.best_index_

print(
    'Средняя MAE на обучающих фолдах:',
    round(-svr_search.cv_results_['mean_train_score'][best_index], 4)
)
print(
    'Средняя MAE на валидационных фолдах:',
    round(-svr_search.best_score_, 4)
)
```

На тестовой выборке LinearSVR достиг MAE 22,26, RMSE 30,32 и R² 0,6517, объясняя около 65,17% вариации ADR. Близкие значения MAE на обучении (22,37), кросс-валидации (22,44) и тесте не указывают на выраженное переобучение.

```python
# Сравнение моделей регрессии
reg_comparison = pd.DataFrame({
    'Прогноз медианой': regression_metrics(
        y_test_reg, dummy_reg_pred
    ),
    'DecisionTreeRegressor': reg_tree_metrics['Test'],
    'LinearSVR': svr_metrics['Test']
})

display(reg_comparison.round(4))
```

Обе модели регрессии превосходят базовый прогноз медианой. Лучшие результаты показало DecisionTreeRegressor: MAE — 13,32, RMSE — 21,35 и R² — 0,8274. У LinearSVR ошибки выше: MAE — 22,26 и RMSE — 30,32, а R² ниже — 0,6517. Поэтому среди исследованных моделей для прогнозирования ADR выбрано дерево регрессии; его преимущество также подтверждается кросс-валидацией.
